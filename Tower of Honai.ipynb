```python
n = int(input("No. of rings = "))
rings = list(map(int, input("Enter ring sizes = ").split(",")))


def position_name(n):
    if 10 < n % 100 < 14:
        return str(n) + "th"

    if n % 10 == 1:
        return str(n) + "st"
    elif n % 10 == 2:
        return str(n) + "nd"
    elif n % 10 == 3:
        return str(n) + "rd"
    else:
        return str(n) + "th"


for i in range(n):
    for j in range(i + 1, n):
        if rings[i] > rings[j]:

            rings[i], rings[j] = rings[j], rings[i]

            print("Move", rings[i], "to",
                  position_name(i + 1), "position")


print("\nFinal Tower:")
for ring in rings:
    print(ring)
```
