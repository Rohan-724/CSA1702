```python
# 31. Program to check whether a given animal is a bird and whether it can fly

birds = {
    "sparrow": True, "eagle": True, "parrot": True, "crow": True,
    "pigeon": True, "swallow": True, "duck": True, "robin": True,
    "peacock": True, "flamingo": True, "woodpecker": True, "owl": True,
    "penguin": False, "ostrich": False, "emu": False, "kiwi": False,
    "cassowary": False, "turkey": False, "hen": False
}

animals = {
    "dog", "cat", "lion", "tiger", "elephant", "horse",
    "cow", "goat", "rabbit", "monkey", "deer", "bear"
}

def check_animal(name):
    if name in birds:
        print("\n", name, "is a bird.")
        if birds[name]:
            print(name, "can fly.")
        else:
            print(name, "cannot fly.")
    elif name in animals:
        print("\n", name, "is not a bird.")
    else:
        print("\n", name, "is not present in the database.")

print("========== BIRD IDENTIFICATION SYSTEM ==========")
print("Enter an animal or bird name to check.")

name = input("\nEnter the name: ").lower().strip()

check_animal(name)
```
