### A* Search Algorithm
The A* search algorithm is a pathfinding algorithm that finds the shortest path between two nodes in a graph. It uses a heuristic function to estimate the cost from the current node to the goal node, combining it with the actual cost from the start node to the current node. This combination helps A* to efficiently explore the most promising paths.

```python
import heapq

def a_star_search(graph, start, goal, heuristic):
    # Priority queue to store (f_cost, current_node, path)
    # f_cost = g_cost + h_cost
    priority_queue = [(0 + heuristic(start, goal), start, [start])]

    # Dictionary to store the lowest g_cost (cost from start) to reach a node
    g_costs = {start: 0}

    # Set to keep track of visited nodes
    visited = set()

    while priority_queue:
        f_cost, current_node, path = heapq.heappop(priority_queue)

        if current_node in visited:
            continue

        visited.add(current_node)

        if current_node == goal:
            return path, g_costs[goal]

        for neighbor, weight in graph.get(current_node, []):
            # Calculate new g_cost for the neighbor
            new_g_cost = g_costs[current_node] + weight

            # If this path to neighbor is better than previous ones, or neighbor not visited yet
            if neighbor not in g_costs or new_g_cost < g_costs[neighbor]:
                g_costs[neighbor] = new_g_cost
                new_f_cost = new_g_cost + heuristic(neighbor, goal)
                heapq.heappush(priority_queue, (new_f_cost, neighbor, path + [neighbor]))

    return None, None # No path found

# --- Example Usage ---

# Define a simple graph (Adjacency List: {node: [(neighbor, weight)]})
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 1)],
    'D': [('G', 3)],
    'E': [('G', 1)],
    'F': [('G', 2)]
}

# Define a heuristic function (straight-line distance for simplicity, replace with actual distances for real maps)
# For this example, we'll use a dummy heuristic (always 0) or a simple dictionary lookup
# A more realistic heuristic would estimate distance to goal based on coordinates or other properties.

# For demonstration, let's create a simple heuristic dictionary for nodes to a 'G' goal
# In a real scenario, this would be calculated dynamically (e.g., Euclidean distance)
heuristic_values = {
    'A': 6, 'B': 4, 'C': 4, 'D': 3, 'E': 1, 'F': 2, 'G': 0
}

def heuristic(node, goal):
    return heuristic_values.get(node, 0)

start_node = 'A'
goal_node = 'G'

path, cost = a_star_search(graph, start_node, goal_node, heuristic)

if path:
    print(f"Shortest path from {start_node} to {goal_node}: {path}")
    print(f"Total cost: {cost}")
else:
    print(f"No path found from {start_node} to {goal_node}")
```
