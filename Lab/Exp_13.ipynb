## Crypto-Arithmetic Problem Solver

Crypto-Arithmetic puzzles involve replacing letters with digits to make a valid arithmetic equation. Each letter must represent a unique digit (0-9). The goal is to find the mapping of letters to digits that satisfies the given equation.

For example:

```
  SEND
+ MORE
------
 MONEY
```

Here, S, E, N, D, M, O, R, Y are unique letters that need to be mapped to unique digits (0-9).

````python
import itertools

def solve_crypto_arithmetic(puzzle):
    """
    Solves a crypto-arithmetic puzzle.

    Args:
        puzzle (str): The puzzle string, e.g., 'SEND + MORE == MONEY'

    Returns:
        dict or None: A dictionary mapping letters to digits if a solution is found,
                      otherwise None.
    """
    words = []
    parts = puzzle.replace(' ', '').split('==')
    equation_parts = [part.split('+') for part in parts]

    for part_list in equation_parts:
        for word in part_list:
            words.append(word)

    # Extract all unique letters
    unique_letters = set("".join(words))
    if len(unique_letters) > 10:
        print("Error: More than 10 unique letters, cannot assign unique digits.")
        return None

    # Get leading letters (cannot be zero)
    leading_letters = set()
    for part_list in equation_parts:
        for word in part_list:
            if word:
                leading_letters.add(word[0])

    # Generate all possible permutations of digits for the unique letters
    for permutation in itertools.permutations(range(10), len(unique_letters)):
        letter_map = dict(zip(unique_letters, permutation))

        # Check leading zeros constraint
        if any(letter_map[letter] == 0 for letter in leading_letters):
            continue

        # Evaluate the expression
        try:
            left_expression = ""
            for i, part in enumerate(equation_parts[0]):
                num_str = "".join(str(letter_map[char]) for char in part)
                left_expression += num_str
                if i < len(equation_parts[0]) - 1:
                    left_expression += " + "

            right_expression = "".join(str(letter_map[char]) for char in equation_parts[1][0])

            if eval(left_expression) == int(right_expression):
                return letter_map
        except KeyError:
            # This should not happen if unique_letters are correctly extracted
            continue
        except SyntaxError:
            # This can happen if a word is empty or invalid, usually caught by earlier checks
            continue

    return None

def display_solution(puzzle, solution):
    if solution:
        print(f"Solution for '{puzzle}':")
        for letter, digit in sorted(solution.items()):
            print(f"  {letter} = {digit}")

        # Substitute letters with digits in the puzzle for verification
        solved_puzzle = puzzle
        for letter, digit in solution.items():
            solved_puzzle = solved_puzzle.replace(letter, str(digit))

        print(f"Verification: {solved_puzzle}")
    else:
        print(f"No solution found for '{puzzle}'.")
````
