```python
from collections import deque

def bfs(graph, start_node):
    """
    Implements Breadth-First Search (BFS) on a given graph.

    Args:
        graph (dict): A dictionary representing the graph where keys are nodes
                      and values are lists of adjacent nodes.
        start_node: The starting node for the BFS traversal.

    Returns:
        list: A list of nodes in the order they were visited by BFS.
    """
    visited = set()
    queue = deque([start_node])
    bfs_traversal = []

    visited.add(start_node)

    while queue:
        current_node = queue.popleft()
        bfs_traversal.append(current_node)

        for neighbor in graph.get(current_node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return bfs_traversal

# --- Example Usage --- #
# Define a sample graph using an adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

start_node = 'A'
breadth_first_traversal = bfs(graph, start_node)

print(f"BFS Traversal starting from node {start_node}: {breadth_first_traversal}")
```
