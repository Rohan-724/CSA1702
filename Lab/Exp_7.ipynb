```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

print(f"Set 1: {set1}")
print(f"Set 2: {set2}")

# Union of two sets
union_set = set1.union(set2)
print(f"Union of Set 1 and Set 2: {union_set}")

# Intersection of two sets
intersection_set = set1.intersection(set2)
print(f"Intersection of Set 1 and Set 2: {intersection_set}")

# Difference of two sets (elements in set1 but not in set2)
difference_set1 = set1.difference(set2)
print(f"Difference (Set 1 - Set 2): {difference_set1}")

# Difference of two sets (elements in set2 but not in set1)
difference_set2 = set2.difference(set1)
print(f"Difference (Set 2 - Set 1): {difference_set2}")

# Symmetric Difference of two sets (elements in either set but not in both)
symmetric_difference_set = set1.symmetric_difference(set2)
print(f"Symmetric Difference: {symmetric_difference_set}")

# Check if one set is a subset of another
print(f"Is {{1, 2}} a subset of Set 1?: {{1, 2}}.issubset(set1)")
print(f"Is Set 1 a subset of {{1, 2}}?: set1.issubset({{1, 2}})")

# Check if one set is a superset of another
print(f"Is Set 1 a superset of {{1, 2}}?: set1.issuperset({{1, 2}})")
print(f"Is {{1, 2}} a superset of Set 1?: {{1, 2}}.issuperset(set1)")
```
