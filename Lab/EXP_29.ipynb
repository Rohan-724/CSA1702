## Prolog Program for PLANETS DB

Below is a simple Prolog program defining some facts about planets. You can copy this code and run it in a Prolog interpreter.

### How to run this Prolog code:
1.  **Install a Prolog Interpreter:** If you don't have one, SWI-Prolog is a popular choice and is free. You can download it from [SWI-Prolog's website](http://www.swi-prolog.org/).
2.  **Save the code:** Save the Prolog code (everything in the ````prolog` block below) into a file named `planets.pl` (or any name with a `.pl` extension).
3.  **Consult the file:** Open your Prolog interpreter and type ````[planets].```` (including the dot) and press Enter. This will load the facts and rules from your file.
4.  **Run queries:** After consulting, you can type the example queries (or your own) at the Prolog prompt.

```prolog
% --- Facts about planets ---

% planet(Name, Type, Distance_from_Sun_AU, Orbital_Period_Earth_Years, Number_of_Moons, Has_Rings).
planet(mercury, terrestrial, 0.39, 0.24, 0, no).
planet(venus, terrestrial, 0.72, 0.62, 0, no).
planet(earth, terrestrial, 1.00, 1.00, 1, no).
planet(mars, terrestrial, 1.52, 1.88, 2, no).
planet(jupiter, gas_giant, 5.20, 11.86, 79, yes).
planet(saturn, gas_giant, 9.58, 29.46, 82, yes).
planet(uranus, ice_giant, 19.22, 84.01, 27, yes).
planet(neptune, ice_giant, 30.05, 164.79, 14, yes).

% --- Relations/Rules ---

% is_inner_planet(Planet): True if the planet is a terrestrial planet.
is_inner_planet(Planet) :-
    planet(Planet, terrestrial, _, _, _, _).

% has_many_moons(Planet): True if the planet has more than 10 moons.
has_many_moons(Planet) :-
    planet(Planet, _, _, _, NumMoons, _),
    NumMoons > 10.

% is_gas_or_ice_giant(Planet): True if the planet is a gas giant or ice giant.
is_gas_or_ice_giant(Planet) :-
    planet(Planet, gas_giant, _, _, _, _).
is_gas_or_ice_giant(Planet) :-
    planet(Planet, ice_giant, _, _, _, _).

% --- Example Queries (you would type these into the Prolog interpreter) ---
% 1. Find all planets:
%    ?- planet(X, _, _, _, _, _).

% 2. Find all terrestrial planets:
%    ?- planet(X, terrestrial, _, _, _, _).

% 3. Which planets have rings?
%    ?- planet(X, _, _, _, _, yes).

% 4. Which planets have no moons?
%    ?- planet(X, _, _, _, 0, _).

% 5. Is Jupiter a terrestrial planet?
%    ?- is_inner_planet(jupiter).

% 6. Which planets have many moons (more than 10)?
%    ?- has_many_moons(X).

% 7. Which planets are gas or ice giants?
%    ?- is_gas_or_ice_giant(X).

% 8. Find the distance from the Sun (in AU) for Mars:
%    ?- planet(mars, _, Distance, _, _, _).

% 9. Find planets with an orbital period less than 5 Earth years:
%    ?- planet(X, _, _, Period, _, _),
%       Period < 5.
```
