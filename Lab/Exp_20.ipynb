## Map Coloring with CSP (Backtracking)

```python
def solve_map_coloring(graph, colors):
    """
    Solves the map coloring problem using a backtracking algorithm.

    Args:
        graph (dict): A dictionary where keys are regions and values are lists of adjacent regions.
        colors (list): A list of available colors.

    Returns:
        dict or None: A dictionary mapping regions to their assigned colors if a solution is found,
                      otherwise None.
    """
    assignment = {}
    regions = list(graph.keys())

    def backtrack():
        # If all regions are assigned a color, a solution is found
        if len(assignment) == len(regions):
            return assignment

        # Select an unassigned region
        unassigned_regions = [r for r in regions if r not in assignment]
        current_region = min(unassigned_regions, key=lambda r: len(get_possible_colors(r)))

        # Try each color for the current region
        for color in colors:
            if is_consistent(current_region, color):
                assignment[current_region] = color
                result = backtrack()
                if result is not None:
                    return result
                del assignment[current_region] # Backtrack: remove assignment
        return None

    def is_consistent(region, color):
        # Check if assigning 'color' to 'region' is consistent with neighbors
        for neighbor in graph.get(region, []):
            if neighbor in assignment and assignment[neighbor] == color:
                return False
        return True

    def get_possible_colors(region):
        # Returns the set of colors that could potentially be assigned to a region
        # based on its already colored neighbors.
        used_colors = set()
        for neighbor in graph.get(region, []):
            if neighbor in assignment:
                used_colors.add(assignment[neighbor])
        return [c for c in colors if c not in used_colors]

    return backtrack()
```

### Example Usage

```python
# Define the map (regions and their adjacencies)
# This example represents a simplified map of Australian states
map_graph = {
    'Western Australia': ['Northern Territory', 'South Australia'],
    'Northern Territory': ['Western Australia', 'South Australia', 'Queensland'],
    'South Australia': ['Western Australia', 'Northern Territory', 'Queensland', 'New South Wales', 'Victoria'],
    'Queensland': ['Northern Territory', 'South Australia', 'New South Wales'],
    'New South Wales': ['South Australia', 'Queensland', 'Victoria'],
    'Victoria': ['South Australia', 'New South Wales'],
    'Tasmania': [] # Tasmania is an island, not adjacent to other states in this simplified map
}

# Define the available colors
available_colors = ['red', 'green', 'blue', 'yellow'] # Using 4 colors as it's often sufficient

print("Attempting to color the map...")
solution = solve_map_coloring(map_graph, available_colors)

if solution:
    print("Map Coloring Solution Found:")
    for region, color in solution.items():
        print(f"{region}: {color}")
else:
    print("No solution found with the given colors and map.")
```
