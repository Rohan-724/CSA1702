```python
def transpose_matrix(matrix):
    """
    Transposes a given matrix.

    Args:
        matrix (list of lists): The input matrix.

    Returns:
        list of lists: The transposed matrix.
    """
    # Handle empty matrix case
    if not matrix or not matrix[0]:
        return []

    num_rows = len(matrix)
    num_cols = len(matrix[0])

    # Create a new matrix with swapped dimensions (cols x rows)
    transposed = [[0 for _ in range(num_rows)] for _ in range(num_cols)]

    # Fill the transposed matrix
    for i in range(num_rows):
        for j in range(num_cols):
            transposed[j][i] = matrix[i][j]
    return transposed

# Example usage:
matrix_a = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

matrix_b = [
    [1, 2],
    [3, 4],
    [5, 6]
]

print("Original Matrix A:")
for row in matrix_a:
    print(row)

transposed_a = transpose_matrix(matrix_a)
print("\nTransposed Matrix A:")
for row in transposed_a:
    print(row)

print("\nOriginal Matrix B:")
for row in matrix_b:
    print(row)

transposed_b = transpose_matrix(matrix_b)
print("\nTransposed Matrix B:")
for row in transposed_b:
    print(row)
```
