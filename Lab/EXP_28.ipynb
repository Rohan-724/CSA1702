```prolog
% Define students
student(john).
student(mary).
student(peter).

% Define teachers
teacher(dr_smith).
teacher(prof_jones).

% Define subjects
subject(math).
subject(physics).
subject(cs).

% Define courses
course(math101, math).
course(phy201, physics).
course(cs300, cs).
course(math202, math).

% Teacher teaches a subject
teaches(dr_smith, math).
teaches(dr_smith, cs).
teaches(prof_jones, physics).

% Student takes a course
takes(john, math101).
takes(john, phy201).
takes(mary, cs300).
takes(peter, math202).
takes(mary, math101).

% Student is enrolled in a subject if they take a course for that subject
enrolled_in(Student, Subject) :-
    takes(Student, Course),
    course(Course, Subject).

% Student has a teacher if they are enrolled in a subject taught by that teacher
has_teacher(Student, Teacher) :-
    enrolled_in(Student, Subject),
    teaches(Teacher, Subject).

% Queries:
% To find all students: student(X).
% To find who teaches math: teaches(Teacher, math).
% To find what courses John takes: takes(john, Course).
% To find what subject math101 is for: course(math101, Subject).
% To find which students are enrolled in math: enrolled_in(Student, math).
% To find John's teachers: has_teacher(john, Teacher).
```

### Explanation of the Prolog Program

This Prolog program defines a knowledge base about students, teachers, subjects, and courses, and their relationships. Here's a breakdown:

*   **Facts:**
    *   `student(Name).`: Declares `Name` as a student (e.g., `student(john).`).
    *   `teacher(Name).`: Declares `Name` as a teacher (e.g., `teacher(dr_smith).`).
    *   `subject(Name).`: Declares `Name` as a subject (e.g., `subject(math).`).
    *   `course(CourseID, Subject).`: Associates a `CourseID` with a `Subject` (e.g., `course(math101, math).`).
    *   `teaches(Teacher, Subject).`: Indicates that a `Teacher` teaches a specific `Subject` (e.g., `teaches(dr_smith, math).`).
    *   `takes(Student, Course).`: Indicates that a `Student` takes a specific `Course` (e.g., `takes(john, math101).`).

*   **Rules:**
    *   `enrolled_in(Student, Subject) :- takes(Student, Course), course(Course, Subject).`:
        This rule defines that a `Student` is `enrolled_in` a `Subject` if they `takes` a `Course`, and that `Course` is for that `Subject`. The `:-` symbol means "if".
    *   `has_teacher(Student, Teacher) :- enrolled_in(Student, Subject), teaches(Teacher, Subject).`:
        This rule defines that a `Student` `has_teacher` `Teacher` if the `Student` is `enrolled_in` a `Subject`, and that `Teacher` `teaches` that `Subject`.

*   **Queries (Examples):**
    The comments at the end of the Prolog code show examples of how you would query this knowledge base in a Prolog interpreter to retrieve information, such as:
    *   `student(X).` to find all defined students.
    *   `teaches(Teacher, math).` to find who teaches mathematics.
    *   `enrolled_in(Student, math).` to find all students enrolled in mathematics.
    *   `has_teacher(john, Teacher).` to find all teachers of John.
