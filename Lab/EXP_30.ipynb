### Towers of Hanoi in Prolog

To run this code, you would typically use a Prolog interpreter (like SWI-Prolog). You can save the code below as a `.pl` file (e.g., `hanoi.pl`) and then load it into the interpreter. To solve the puzzle, you would then query `hanoi(N, left, center, right).` where `N` is the number of discs.

```python
%%writefile hanoi.pl

% Base case: Moving 0 discs requires no moves.
hanoi(0, _, _, _) :- !.

% Recursive case: Moving N discs from Start to End using Aux
hanoi(N, Start, Aux, End) :-
    M is N - 1,
    hanoi(M, Start, End, Aux),       % Move N-1 discs from Start to Aux (using End as temp)
    format('Move disc ~w from ~w to ~w~n', [N, Start, End]), % Move the Nth disc from Start to End
    hanoi(M, Aux, Start, End).       % Move N-1 discs from Aux to End (using Start as temp)


% Example usage:
% To run this, save the content as 'hanoi.pl'.
% Then in a Prolog interpreter (like SWI-Prolog):
% ?- ['hanoi.pl'].
% % hanoi.pl compiled 0.00 sec, 2 clauses
% ?- hanoi(3, left, center, right).
% Move disc 1 from left to right
% Move disc 2 from left to center
% Move disc 1 from right to center
% Move disc 3 from left to right
% Move disc 1 from center to left
% Move disc 2 from center to right
% Move disc 1 from left to right
% true.
```
