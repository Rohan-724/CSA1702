```python
from collections import deque

def water_jug_problem(jug1_capacity, jug2_capacity, target):
    """
    Solves the Water Jug Problem using Breadth-First Search (BFS).

    Args:
        jug1_capacity (int): The capacity of the first jug.
        jug2_capacity (int): The capacity of the second jug.
        target (int): The target amount of water to measure.

    Returns:
        list: A list of steps to reach the target, or None if no solution exists.
    """

    # Use a deque for efficient BFS operations (pop from left, append to right)
    queue = deque()
    # Store visited states to avoid redundant computations and cycles
    visited = set()
    # Store the path to reconstruct the solution
    path = []

    # Initial state: both jugs are empty. (jug1_current, jug2_current, history_of_moves)
    initial_state = (0, 0, [])
    queue.append(initial_state)
    visited.add((0, 0))

    while queue:
        current_jug1, current_jug2, current_path = queue.popleft()

        # Check if the target is reached in either jug
        if current_jug1 == target or current_jug2 == target:
            return current_path + [(current_jug1, current_jug2, "Target Reached!")]

        # --- Generate possible next states (operations) ---

        # 1. Fill Jug 1
        if current_jug1 < jug1_capacity:
            next_state = (jug1_capacity, current_jug2)
            if next_state not in visited:
                visited.add(next_state)
                queue.append((next_state[0], next_state[1], current_path + [(next_state[0], next_state[1], f"Fill Jug 1 (from {current_jug1}) -> Jug1:{next_state[0]}, Jug2:{next_state[1]}")]))

        # 2. Fill Jug 2
        if current_jug2 < jug2_capacity:
            next_state = (current_jug1, jug2_capacity)
            if next_state not in visited:
                visited.add(next_state)
                queue.append((next_state[0], next_state[1], current_path + [(next_state[0], next_state[1], f"Fill Jug 2 (from {current_jug2}) -> Jug1:{next_state[0]}, Jug2:{next_state[1]}")]))

        # 3. Empty Jug 1
        if current_jug1 > 0:
            next_state = (0, current_jug2)
            if next_state not in visited:
                visited.add(next_state)
                queue.append((next_state[0], next_state[1], current_path + [(next_state[0], next_state[1], f"Empty Jug 1 (from {current_jug1}) -> Jug1:{next_state[0]}, Jug2:{next_jug2}")]))

        # 4. Empty Jug 2
        if current_jug2 > 0:
            next_state = (current_jug1, 0)
            if next_state not in visited:
                visited.add(next_state)
                queue.append((next_state[0], next_state[1], current_path + [(next_state[0], next_state[1], f"Empty Jug 2 (from {current_jug2}) -> Jug1:{next_state[0]}, Jug2:{next_state[1]}")]))

        # 5. Pour Jug 1 to Jug 2
        if current_jug1 > 0 and current_jug2 < jug2_capacity:
            pour_amount = min(current_jug1, jug2_capacity - current_jug2)
            next_jug1 = current_jug1 - pour_amount
            next_jug2 = current_jug2 + pour_amount
            next_state = (next_jug1, next_jug2)
            if next_state not in visited:
                visited.add(next_state)
                queue.append((next_state[0], next_state[1], current_path + [(next_state[0], next_state[1], f"Pour Jug 1 to Jug 2 ({pour_amount} units) -> Jug1:{next_state[0]}, Jug2:{next_state[1]}")]))

        # 6. Pour Jug 2 to Jug 1
        if current_jug2 > 0 and current_jug1 < jug1_capacity:
            pour_amount = min(current_jug2, jug1_capacity - current_jug1)
            next_jug1 = current_jug1 + pour_amount
            next_jug2 = current_jug2 - pour_amount
            next_state = (next_jug1, next_jug2)
            if next_state not in visited:
                visited.add(next_state)
                queue.append((next_state[0], next_state[1], current_path + [(next_state[0], next_state[1], f"Pour Jug 2 to Jug 1 ({pour_amount} units) -> Jug1:{next_state[0]}, Jug2:{next_state[1]}")]))

    return None  # No solution found

# --- Example Usage ---
```

```python
# Example 1: 4-gallon jug, 3-gallon jug, target 2 gallons
jug1_capacity_ex1 = 4
jug2_capacity_ex1 = 3
target_ex1 = 2

solution_ex1 = water_jug_problem(jug1_capacity_ex1, jug2_capacity_ex1, target_ex1)

print(f"\nWater Jug Problem: Jug1 Capacity={jug1_capacity_ex1}, Jug2 Capacity={jug2_capacity_ex1}, Target={target_ex1}")
if solution_ex1:
    for step_num, (j1, j2, desc) in enumerate(solution_ex1):
        print(f"Step {step_num + 1}: ({j1}, {j2}) - {desc}")
else:
    print("No solution found.")

# Example 2: 5-gallon jug, 3-gallon jug, target 4 gallons
jug1_capacity_ex2 = 5
jug2_capacity_ex2 = 3
target_ex2 = 4

solution_ex2 = water_jug_problem(jug1_capacity_ex2, jug2_capacity_ex2, target_ex2)

print(f"\nWater Jug Problem: Jug1 Capacity={jug1_capacity_ex2}, Jug2 Capacity={jug2_capacity_ex2}, Target={target_ex2}")
if solution_ex2:
    for step_num, (j1, j2, desc) in enumerate(solution_ex2):
        print(f"Step {step_num + 1}: ({j1}, {j2}) - {desc}")
else:
    print("No solution found.")

# Example 3: 6-gallon jug, 4-gallon jug, target 5 gallons (impossible)
jug1_capacity_ex3 = 6
jug2_capacity_ex3 = 4
target_ex3 = 5

solution_ex3 = water_jug_problem(jug1_capacity_ex3, jug2_capacity_ex3, target_ex3)

print(f"\nWater Jug Problem: Jug1 Capacity={jug1_capacity_ex3}, Jug2 Capacity={jug2_capacity_ex3}, Target={target_ex3}")
if solution_ex3:
    for step_num, (j1, j2, desc) in enumerate(solution_ex3):
        print(f"Step {step_num + 1}: ({j1}, {j2}) - {desc}")
else:
    print("No solution found. (Note: This specific target is impossible with these jug sizes because 5 is not a multiple of gcd(6, 4)=2)")
```
