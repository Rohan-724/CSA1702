### Traveling Salesman Problem (TSP) Implementation

The Traveling Salesman Problem (TSP) is an NP-hard problem in combinatorial optimization. Given a list of cities and the distances between each pair of cities, the problem is to find the shortest possible route that visits each city exactly once and returns to the origin city.

Here, we'll implement a solution using dynamic programming, which is an exact method suitable for a small number of cities (typically up to 15-20 cities).

```python
import math

def solve_tsp_dp(distances):
    """
    Solves the Traveling Salesman Problem using dynamic programming.

    Args:
        distances (list of lists): A square matrix where distances[i][j]
                                   is the distance between city i and city j.

    Returns:
        tuple: A tuple containing:
            - float: The minimum tour cost.
            - list: The optimal tour (list of city indices).
    """
    n = len(distances)
    # dp[mask][i] will be the minimum cost to visit cities in 'mask'
    # ending at city 'i'.
    dp = [[math.inf] * n for _ in range(1 << n)]

    # parent[mask][i] stores the predecessor city in the optimal path
    parent = [[-1] * n for _ in range(1 << n)]

    # Base case: starting at city 0, visiting only city 0
    dp[1][0] = 0

    # Iterate through all possible masks (subsets of cities)
    for mask in range(1, 1 << n):
        for i in range(n):
            # If city i is in the current mask
            if (mask >> i) & 1:
                # Consider all possible previous cities j
                for j in range(n):
                    if i == j: # Cannot go from i to i
                        continue
                    # If city j is in the mask and not i, and going from j to i is valid
                    if (mask >> j) & 1 and distances[j][i] != math.inf:
                        prev_mask = mask ^ (1 << i) # Mask without city i
                        if dp[prev_mask][j] + distances[j][i] < dp[mask][i]:
                            dp[mask][i] = dp[prev_mask][j] + distances[j][i]
                            parent[mask][i] = j

    # Find the minimum cost to return to the starting city (city 0)
    min_cost = math.inf
    last_city = -1
    final_mask = (1 << n) - 1 # All cities visited
    for i in range(1, n):
        if dp[final_mask][i] + distances[i][0] < min_cost:
            min_cost = dp[final_mask][i] + distances[i][0]
            last_city = i

    # Reconstruct the path
    tour = []
    current_mask = final_mask
    current_city = last_city
    while current_city != -1:
        tour.append(current_city)
        prev_city = parent[current_mask][current_city]
        current_mask ^= (1 << current_city) # Remove current_city from mask
        current_city = prev_city

    # The previous loop reconstructs the path from last_city back to city 0 (including 0).
    # Now, reverse it to get 0 -> ... -> last_city
    tour.reverse()

    # Add city 0 to complete the cycle (0 -> ... -> last_city -> 0)
    tour.append(0)

    return min_cost, tour
```

### Example Usage

```python
# Example: 4 cities
# distances[i][j] is the distance from city i to city j
cities = ['A', 'B', 'C', 'D']
distances = [
    [0, 10, 15, 20],  # A to A, B, C, D
    [10, 0, 35, 25],  # B to A, B, C, D
    [15, 35, 0, 30],  # C to A, B, C, D
    [20, 25, 30, 0]   # D to A, B, C, D
]

min_cost, optimal_tour_indices = solve_tsp_dp(distances)

# Convert numerical indices to city names for better readability
optimal_tour_cities = [cities[i] for i in optimal_tour_indices]

print(f"Minimum tour cost: {min_cost}")
print(f"Optimal tour: {optimal_tour_cities}")

print("\n--- Another Example (5 cities) ---")
cities_2 = ['X', 'Y', 'Z', 'W', 'V']
distances_2 = [
    [0, 2, 4, 1, 5], # X
    [2, 0, 3, 2, 4], # Y
    [4, 3, 0, 5, 2], # Z
    [1, 2, 5, 0, 3], # W
    [5, 4, 2, 3, 0]  # V
]

min_cost_2, optimal_tour_indices_2 = solve_tsp_dp(distances_2)
optimal_tour_cities_2 = [cities_2[i] for i in optimal_tour_indices_2]

print(f"Minimum tour cost: {min_cost_2}")
print(f"Optimal tour: {optimal_tour_cities_2}")
```
