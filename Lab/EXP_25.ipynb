```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np

# 1. Define the Feedforward Neural Network Model
# We'll create a simple sequential model for demonstration

def create_ffnn_model(input_shape, num_classes):
    model = keras.Sequential([
        layers.Input(shape=input_shape), # Input layer
        layers.Dense(128, activation='relu', name='hidden_layer_1'), # First hidden layer
        layers.Dense(64, activation='relu', name='hidden_layer_2'),  # Second hidden layer
        layers.Dense(num_classes, activation='softmax', name='output_layer') # Output layer
    ])
    return model

# Input features (e.g., 10 features per sample)
x_train = np.random.rand(100, 10).astype(np.float32)
y_train = np.random.randint(0, 2, 100).astype(np.float32) # Binary classification (0 or 1)


# For simplicity, let's assume y_train are class indices (0 or 1) and we use SparseCategoricalCrossentropy.
num_classes_example = 2

# 3. Create, Compile, and Summarize the Model
input_dim = x_train.shape[1]
model = create_ffnn_model(input_shape=(input_dim,), num_classes=num_classes_example)

model.compile(
    optimizer='adam',
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=False),
    metrics=['accuracy']
)

print("\nModel Summary:")
model.summary()

# 4. Train the Model
print("\nTraining the model...")
history = model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=32,
    verbose=1
)

print("\nTraining complete!")

# 5. Make a prediction with the trained model
print("\nMaking a prediction:")
sample_input = np.random.rand(1, 10).astype(np.float32)
predictions = model.predict(sample_input)
print(f"Sample Input: {sample_input}")
print(f"Predicted Probabilities: {predictions}")
print(f"Predicted Class: {np.argmax(predictions)}")
```
