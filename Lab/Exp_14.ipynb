```python
def solve_missionaries_cannibals():
    # State representation: (M_left, C_left, boat_pos, M_right, C_right)
    # M = Missionaries, C = Cannibals
    # boat_pos: 0 for left bank, 1 for right bank
    initial_state = (3, 3, 0, 0, 0)
    goal_state = (0, 0, 1, 3, 3)

    # Define valid moves: (missionaries_to_move, cannibals_to_move)
    # Boat capacity is 2
    moves = [
        (1, 0),  # 1 missionary
        (2, 0),  # 2 missionaries
        (0, 1),  # 1 cannibal
        (0, 2),  # 2 cannibals
        (1, 1)   # 1 missionary and 1 cannibal
    ]

    # Queue for BFS: stores (state, path_to_state)
    queue = [(initial_state, [initial_state])]
    visited = {initial_state}

    while queue:
        current_state, path = queue.pop(0)

        if current_state == goal_state:
            return path

        M_left, C_left, boat_pos, M_right, C_right = current_state

        for dM, dC in moves:
            if boat_pos == 0:  # Boat is on the left bank, moving to right
                new_M_left, new_C_left = M_left - dM, C_left - dC
                new_M_right, new_C_right = M_right + dM, C_right + dC
                new_boat_pos = 1
            else:  # Boat is on the right bank, moving to left
                new_M_left, new_C_left = M_left + dM, C_left + dC
                new_M_right, new_C_right = M_right - dM, C_right - dC
                new_boat_pos = 0

            next_state = (new_M_left, new_C_left, new_boat_pos, new_M_right, new_C_right)

            # Check for validity of the next state
            if (0 <= new_M_left <= 3 and 0 <= new_C_left <= 3 and
                0 <= new_M_right <= 3 and 0 <= new_C_right <= 3):

                # Check if cannibals outnumber missionaries on either bank (unless 0 missionaries)
                if (new_M_left == 0 or new_M_left >= new_C_left) and \
                   (new_M_right == 0 or new_M_right >= new_C_right):

                    if next_state not in visited:
                        visited.add(next_state)
                        queue.append((next_state, path + [next_state]))
    return None

def print_solution(path):
    if not path:
        print("No solution found.")
        return

    print("Solution for Missionaries and Cannibals Problem:")
    print("--------------------------------------------------")
    for i, state in enumerate(path):
        M_left, C_left, boat_pos, M_right, C_right = state
        boat_side = "left" if boat_pos == 0 else "right"
        print(f"Step {i}: Left Bank (M: {M_left}, C: {C_left}) | Boat on {boat_side} | Right Bank (M: {M_right}, C: {C_right})")
    print("--------------------------------------------------")

# Run the solver
solution_path = solve_missionaries_cannibals()
print_solution(solution_path)
```
