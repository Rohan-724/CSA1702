## Python List Methods

### 1. `append()` Method (Add a single element to the end)

```python
my_list = [1, 2, 3]
print(f"Original list: {my_list}")

my_list.append(4)
print(f"After append(4): {my_list}")

my_list.append('hello')
print(f"After append('hello'): {my_list}")
```

### 2. `extend()` Method (Add multiple elements from an iterable to the end)

```python
my_list = [1, 2, 3]
print(f"Original list: {my_list}")

my_list.extend([4, 5, 6])
print(f"After extend([4, 5, 6]): {my_list}")

my_list.extend(('a', 'b'))
print(f"After extend(('a', 'b')): {my_list}")
```

### 3. Adding Lists (Concatenation)

```python
list1 = [1, 2]
list2 = [3, 4]

new_list = list1 + list2
print(f"list1: {list1}")
print(f"list2: {list2}")
print(f"Concatenated list (list1 + list2): {new_list}")
```

### 4. Deleting Elements

#### a. `del` statement (Delete by index or slice)

```python
my_list = ['a', 'b', 'c', 'd', 'e']
print(f"Original list: {my_list}")

del my_list[1] # Delete element at index 1 ('b')
print(f"After del my_list[1]: {my_list}")

del my_list[1:3] # Delete elements from index 1 to 2 (['c', 'd'])
print(f"After del my_list[1:3]: {my_list}")
```

#### b. `pop()` method (Remove and return element by index, or last element if no index is given)

```python
my_list = [10, 20, 30, 40, 50]
print(f"Original list: {my_list}")

removed_element = my_list.pop(2) # Remove element at index 2 (30)
print(f"After pop(2): {my_list}, Removed element: {removed_element}")

last_element = my_list.pop() # Remove the last element (50)
print(f"After pop(): {my_list}, Last element removed: {last_element}")
```

#### c. `remove()` method (Remove first occurrence of a specified value)

```python
my_list = ['apple', 'banana', 'cherry', 'banana', 'date']
print(f"Original list: {my_list}")

my_list.remove('banana') # Remove the first 'banana'
print(f"After remove('banana'): {my_list}")

# If the item is not found, it raises a ValueError
try:
    my_list.remove('grape')
except ValueError as e:
    print(f"Error: {e}")
```
