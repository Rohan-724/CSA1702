Here's a Prolog program to calculate the sum of integers from 1 to `N`. This program uses a recursive approach:

```prolog
% Base case: The sum from 1 to 0 is 0.
sum_up_to(0, 0).

% Recursive case: To sum up to N, add N to the sum up to N-1.
% N > 0 is a guard to ensure N is positive.
sum_up_to(N, Sum) :-
    N > 0,
    N_minus_1 is N - 1,
    sum_up_to(N_minus_1, Sum_N_minus_1),
    Sum is N + Sum_N_minus_1.
```

**How to use it (in a Prolog interpreter):**

To find the sum of integers up to, say, 5, you would query:

```prolog
?- sum_up_to(5, Result).
```

This would output `Result = 15`.
