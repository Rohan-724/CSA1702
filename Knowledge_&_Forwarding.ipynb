```python
# Knowledge Representation using Python

# --------------------------m
# Facts
# --------------------------
facts = {"Cat(Tom)"}

# --------------------------
# Rules
# Key = Conclusion
# Value = Required Fact
# --------------------------
rules = {
    "Mammal(Tom)": "Cat(Tom)",
    "Animal(Tom)": "Mammal(Tom)"
}

print("Initial Facts:")
for fact in facts:
    print(fact)

print("\nApplying Rules...")

# Keep applying rules until no new facts are added
changed = True

while changed:
    changed = False

    for conclusion, premise in rules.items():

        if premise in facts and conclusion not in facts:
            facts.add(conclusion)
            print(f"Derived: {conclusion}")
            changed = True

print("\nFinal Knowledge Base:")
for fact in facts:
    print(fact)
```

```python
# Forward Chaining Example

# Known facts
facts = {"Human(Socrates)"}

# Rule: IF Human(Socrates) THEN Mortal(Socrates)# Forward Chaining Example

# Initial fact
facts = {"Didn't study"}

# Rule: If Study then Pass
if "Study" in facts:
    facts.add("Pass")
else:
  facts.add("Fail")

# Print inferred facts
print("Knowledge Base:")
for fact in facts:
    print(fact)
if "Human(Socrates)" in facts:
    facts.add("Mortal(Socrates)")

# Display all facts after inference
print("Facts after Forward Chaining:")
for fact in facts:
    print(fact)
```
