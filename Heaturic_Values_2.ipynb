```python
# Cost from Start
g = {
    'A': 2,
    'B': 4,
    'C': 3
}

# Heuristic to Goal
h = {
    'A': 6,
    'B': 2,
    'C': 3
}

print("Node\tg(n)\th(n)\tf(n)")

f = {}

for node in g:
    f[node] = g[node] + h[node]
    print(node, "\t", g[node], "\t", h[node], "\t", f[node])

best = min(f, key=f.get)

print("\nBest Node to Visit:", best)
A* algorithm
```
